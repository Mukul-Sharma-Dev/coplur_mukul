```python
import numpy as np
arr = np.array([[6, -8, 73, -110],
                [np.nan, -8, 0, 94]])
arr = np.nan_to_num(arr, nan=0)

print("Array after replacing NaN with 0:")
print(arr)
arr_transpose = arr.T
print("\nAfter interchanging rows and columns:")
print(arr_transpose)
```

```python
import numpy as np
arr = np.array([[10, np.nan, 30],
                [20, 40, np.nan],
                [30, 50, 60]])
col_mean = np.nanmean(arr, axis=0)
inds = np.where(np.isnan(arr))
arr[inds] = np.take(col_mean, inds[1])
print("Array after replacing NaN with column averages:")
print(arr)
```

```python
arr = np.array([10, -5, 20, -8, 15])

result = np.where(arr < 0, 0, arr)

print(result)
```

```python
import numpy as np
from scipy import stats
arr1 = np.array([[3, 4],
                 [5, 6]])
arr2 = np.array([[1, 0],
                 [7, 8]])
avg = (arr1 + arr2) / 2

print("Average Array:")
print(avg)

combined = np.concatenate((arr1.flatten(), arr2.flatten()))
print("\nMean =", np.mean(combined))
print("Median =", np.median(combined))
mode_result = stats.mode(combined, keepdims=False)
print("Mode =", mode_result.mode)
```

```python
import numpy as np

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])
solution = np.linalg.solve(A, B)

print("Solution using linalg.solve():")
print("x =", solution[0])
print("y =", solution[1])
print("z =", solution[2])
```

```python
#using inverse matrix
import numpy as np

A = np.array([[1, -2, 3],
              [-1, 3, -1],
              [2, -5, 5]])

B = np.array([9, -6, 17])
A_inv = np.linalg.inv(A)
X = np.dot(A_inv, B)
print("Solution using Inverse Matrix Method:")
print("x =", X[0])
print("y =", X[1])
print("z =", X[2])
```

```python
import matplotlib.pyplot as plt
import numpy as np
subjects = ['Math', 'Physics', 'Chemistry', 'English', 'CS']

sem1 = [75, 80, 70, 85, 90]
sem2 = [82, 78, 88, 84, 95]

x = np.arange(len(subjects))

plt.figure(figsize=(8,5))

plt.plot(x, sem1, marker='o', linewidth=2, label='Semester 1')
plt.plot(x, sem2, marker='s', linewidth=2, label='Semester 2')

plt.title('Semester Result Comparison')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.xticks(x, subjects)
plt.grid(True)
plt.legend()

plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np

subjects = ['Math', 'Physics', 'Chemistry', 'English', 'CS']

sem1 = [75, 80, 70, 85, 90]
sem2 = [82, 78, 88, 84, 95]

x = np.arange(len(subjects))
width = 0.35

plt.figure(figsize=(8,5))

plt.bar(x-width/2, sem1, width, label='Semester 1')
plt.bar(x+width/2, sem2, width, label='Semester 2')

plt.title('Semester Result Comparison')
plt.xlabel('Subjects')
plt.ylabel('Marks')
plt.xticks(x, subjects)
plt.legend()

plt.show()
```
